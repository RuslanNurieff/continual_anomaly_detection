```python
from moviad.datasets.bmad.bmad_dataset import BMAD
from moviad.models.draem.augmentation import DRAEMTrain
from moviad.datasets.mvtec.mvtec_dataset import MVTecDataset

bmad_train = BMAD("segmentation", "/mnt/disk1/ruslan_nuriev/bmad", "liver", "train", norm=False)
# bmad_train.load_dataset()
draem_train = DRAEMTrain(bmad_train, "/mnt/disk1/manuel_barusco/CL_VAD/adcl_paper/anomaly_dataset/images", split="train", resize_shape=[224, 224])

bmad_test = BMAD("segmentation", "/mnt/disk1/ruslan_nuriev/bmad", "liver", "test", norm=False)
# bmad_test.load_dataset()
draem_test = DRAEMTrain(bmad_test, "/mnt/disk1/manuel_barusco/CL_VAD/adcl_paper/anomaly_dataset/images", split="test", resize_shape=[224, 224])
```

```python
import torch

draem_train = torch.utils.data.DataLoader(draem_train, batch_size=8)
draem_test = torch.utils.data.DataLoader(draem_test, batch_size=8)
```

```python
from moviad.models.draem.draem import DRAEM
from moviad.trainers.trainer_draem import TrainerDRAEM

model = DRAEM("cuda:0")
trainer = TrainerDRAEM(model, draem_train, draem_test, "cuda:0")
```
